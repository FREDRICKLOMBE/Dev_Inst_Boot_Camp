```python
#1 SERIES: A one-dimensional array-like object capable of holding any data type.
# Think of it as a column in a spreadsheet.
data_1 = pd.Series([1, 3, 5, 7, 9])
```

```python
# DataFrames: A two-dimensional, size-mutable, and potentially heterogeneous tabular data structure 
# with labeled axes (rows and columns). It’s like a spreadsheet or SQL table.

data = {
    'Name': ['John', 'Anna', 'Peter', 'Linda'],
    'Age': [28, 34, 29, 32],
    'City': ['New York', 'Paris', 'Berlin', 'London']
}
df = pd.DataFrame(data)
```

```python
# Filtering: Select rows based on a certain condition.
df[df['Age'] >= 30]         # Selects rows where age is greater than 30
```

```python
# Filters the DataFrame to people aged 30 or older, then selects only their Name
df[df['Age'] >= 30]['Name']
```

```python
# Sorting: Sort data by a specific column.
sorted_df = df.sort_values(by="Name")
sorted_df
```
