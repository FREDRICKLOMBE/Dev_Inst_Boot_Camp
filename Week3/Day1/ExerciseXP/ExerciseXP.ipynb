****🌟Exercise 1: Introduction to Data Analysis****

**1. What Is Data Analysis?**

Data analysis is the systematic process of collecting, cleaning, organizing, transforming, and examining data with a sole purpose of discovering useful information, patterns, and meaningful insights that can support better decisions. For example, a business may have thousands of records containing customer purchases. By analyzing this data, the business can determine which products are most popular, when customers are most likely to buy, and which products are underperforming. Instead of relying on assumptions, managers can use evidence from the data to make informed decisions.
Data analysis can also be divided into different approaches. Descriptive analysis examines what happened, diagnostic analysis investigates why something happened, predictive analysis estimates what may happen in the future, and prescriptive analysis helps determine what action should be taken.

**2. Why Is Data Analysis Important in Modern Contexts?**

Ultimately, the value of data does not come simply from having large amounts of information. Its value comes from being able to analyze that information effectively and transform it into reliable insights that can improve decisions and solve real-world problems.

One of the main benefits of data analysis is better decision-making. Instead of making decisions based primarily on assumptions or intuition, organizations can use evidence and measurable trends. For example, a company can analyze sales data to determine which products should receive more investment.
Data analysis also helps organizations identify problems and improve efficiency. An organization can examine its operations to identify delays, unnecessary costs, declining performance, or areas where resources are being wasted. In healthcare, for example, analyzing patient and population data can support treatment decisions, disease monitoring, and long-term planning.
Another important benefit is prediction and planning. Organizations can use historical data together with statistical models and machine learning to forecast demand, identify risks, and anticipate future trends.

**3. Three Areas Where Data Analysis Is Applied Today**

**In Environmental management and Agriculture:** Environmental managers and farmers can analyze information about rainfall, temperature, soil conditions, water quality, biodiversity, land use, and vegetation to understand environmental changes and support sustainable resource management. For example, satellite imagery and GIS data can be analyzed to monitor changes in forest cover, identify areas affected by drought, assess agricultural productivity, and monitor environmental degradation. Combining these data sources with statistical analysis and machine learning can help decision-makers develop more effective environmental and agricultural strategies.

**In Hospitals and healthcare** organizations collect information from electronic health records, medical tests, patient surveys, wearable devices, and public health databases. Analyzing this information can help healthcare professionals identify health risks, support diagnosis and treatment decisions, monitor diseases, and allocate resources. At a population level, health authorities can analyze disease patterns and vaccination data to understand public health trends and plan interventions. For example, predictive models can analyze patient information to identify people who may be at high risk of developing complications. This allows healthcare professionals to be proactive than reactive.

**In Businesses** data analysis is used to understand customers, monitor performance, forecast sales, manage inventory, and improve marketing strategies. For example, an online retailer can analyze customers' purchasing histories to identify popular products and recommend relevant products to individual customers. Entrepreneurs can also use data analysis to identify market opportunities. By studying customer demographics, purchasing behavior, competitors, and market trends, entrepreneurs can make better decisions about what products or services to develop.

🌟 Exercise 6: Identifying Data Types\
Below are various data sources. Identify whether each one is an example of structured or unstructured data.


| Data Source | Structured / Unstructured | Why |
|---|---|---|
| Company's financial reports in an Excel file | Structured | Data sits in rows and columns with defined fields (dates, amounts, categories), easy to search, sort, and analyze directly. |
| Photographs uploaded to a social media platform | Unstructured | Images have no predefined fields or tabular format; extracting meaning requires specialized processing like computer vision. |
| A collection of news articles on a website | Unstructured | Free-form text with no fixed schema separating things like sentiment, topic, or key facts. |
| Inventory data in a relational database | Structured | Organized into predefined tables with rows and columns (SKU, quantity, price, location), following a strict schema. |
| Recorded interviews from a market research study | Unstructured | Audio/spoken content with no fixed structure; requires transcription and further processing to extract usable information. |

🌟 Exercise 9 : Export a dataframe to excel format and JSON format.

```python
import numpy as np
import pandas as pd
from faker import Faker

# Create a Faker object
fake = Faker()

# List of fruits
fruits = [
    "Apple",
    "Banana",
    "Orange",
    "Mango",
    "Pineapple",
    "Watermelon",
    "Grapes",
    "Strawberry",
    "Papaya",
    "Avocado"
]

# Generate 100 synthetic sales records
data = []

for _ in range(100):
    product_name = fake.random.choice(fruits)
    sale_amount = np.random.uniform(5, 10)
    date_of_sale = fake.date_between(start_date='-2y', end_date='today')
    customer_age = np.random.randint(18, 70)

    data.append([
        product_name,
        sale_amount,
        date_of_sale,
        customer_age
    ])

# Convert the data into a Pandas DataFrame
sales_df = pd.DataFrame(
    data,
    columns=['Product Name', 'Sale Amount', 'Date of Sale', 'Customer Age']
)

# Display the first 5 rows
print(sales_df.head())
```

```python
#Export to JSON
sales_df.to_json("sales_dataframe.json")

#Export to xlsx
sales_df.to_excel("sales_dataframe.xlsx", sheet_name = "Sales_Records")
```
