```python
import pandas as pd

#0Create a DataFrame

df = pd.DataFrame({
    "Column1": [1, 2, 3, 4, 5],
    "Column2": ["A", "B", "C", "D", "E"],
    "Column3": [True, False, True, False, True]

})

#Export to csv
df.to_csv("my_dataframe.csv", index = False)

#Export to JSON
df.to_json("my_dataframe.json")

#Export to xlsx
df.to_excel("my_dataframe.xlsx", sheet_name = "random")
```
